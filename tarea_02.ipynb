```python
# Problema 1: Clasificador de temperatura
# Recorre los números desde -30 hasta 50 inclusive (por eso range llega a 51)
for temp in range(-30, 51):
    if temp < 0:
        print(f"{temp} Muy frío")
    elif 0 <= temp <= 24:
        print(f"{temp} Templado")
    else:
        # Mayor o igual a 25
        print(f"{temp} Caluroso")
```

```python
# Problema 2: Números especiales del robot
# Recorre los números del 1 al 100
for num in range(1, 101):
    # Condición principal: que sea divisible por 5 (resto igual a 0)
    if num % 5 == 0:
        # Si además es divisible por 10, muestra "Bip Bip"
        if num % 10 == 0:
            print(f"{num} Bip Bip")
        else:
            print(f"{num} Bip")
```

# Problema 3: Detector de puntaje gamer
# Solicita dos números enteros al usuario
puntaje_1 = int(input("Ingresa primer puntaje: "))
puntaje_2 = int(input("Ingresa segundo puntaje: "))

suma = puntaje_1 + puntaje_2

# Clasificación según el resultado de la suma
if suma < 50:
    print("Resultado: Nivel principiante")
elif 50 <= suma < 80:
    print("Resultado: Nivel intermedio")
else:
    print("Resultado: Nivel pro")
